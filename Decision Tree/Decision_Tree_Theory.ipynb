## DECISION TREE

What is Decision Tree?

- Decision tree is a supervised machine learning algorithm that makes predictions by asking series of questions.

- It learns by asking the best possible questions until reaches an answer.

- Decision tree algorithm is used for both classification and regression problems.

- whether we are Hungry to help you arrive at one of two final decisions: Walk or Bus.

So this way we can solve classification and regression problems using Decision tree algorithm.

Why is it Called a Tree?

- It looks like an up side down tree including with root and branches.




---


*Applications*

- Healthcare -> Disease prediction based on age, blood pressure.

- Bank -> Loan approval based on Income, credit score.

- HR -> Employee Attrition based on Salary and experience.

- Education -> Student performance based on Attendance and assignements.

- E-Commerce -> Will customer buy based on income and Age.




---



*Advantages*

- Easy to understand looks like human thinking

- Works with numbers and categories (Regression and classification problems).

- Little data preparation, no need of normalization and scaling



---



*Limitations*

- Overfitting -> the tree can memorize the training data.

- Can become very large -> Deep trees are hard to interpret.

- Sensitive to small changes -> Changing a few rows can produce a different tree.

**IMPORTANT TERMINOLOGIES IN DECISION TREE**

We can understand the terminalogies with below example.

Problem statement : Will a student PASS or FAIL?

| Student | Study Hours | Attendance | Result |
| :--- | :--- | :--- | :--- |
| A | 6 | 90% | Pass |
| B | 2 | 60% | Fail |
| C | 5 | 95% | Pass |
| D | 3 | 70% | Fail |
| E | 7 | 92% | Pass |

***Decision Tree***

  ```text
                Study Hours > 4?
                     │
             Yes ────┴──── No
              │             │
     Attendance > 80?      Fail
         │
    Yes ─┴── No
     │        │
   Pass     Fail
```

**- Root Node**

Root node is the first question the decision tree asks. It is starting point of any decision tree.

Here, `Study Hours > 4?` is the Root node.





---



**- Decision Node**

Decision node the next question after the Root node.

Here, `Attendance > 80%?` is the decison node.

Decision code is also known as child node.



---



**- Leaf Node**

Leaf Node is the final node of the decision tree and after than there is no more questions.

Here, `Pass` and `Fail` are the Leaf nodes.



---



**- Branch**

Branch connects one node to the other and usually represents as `Yes` and `No`.

Here, `Study Hours >4?` followed by `Yes` and `No`, these are branches.



---



**- Splitting**

Splitting means dividing data into smaller groups.

For example, if we are dividing 20 students, after asking `Study Hours >4?` with `Yes` by 12 students and `No` by 8 students.

This process is called as splitting.



---



**- Pruning**

Pruning is the process of cutting the unnecessary branches.

*Why do we use pruning?*

Suppose if the tree becomes very large due to more number of unnecessary features. Instead of learning, it starts memorizing the data. It remembers everything and it is overfitting the data.

To solve it, we cut unnecessary branches. This is called Pruning.



---



**- Depth**

Depth measures from Root node to leaf node. How many levels deep the tree goes.

Here, Level 1-> Study hours

Level 2 -> Attendance

Level 3 -> Pass

Depth here is 3.




---



**- Pure Node**

If the tree has only one split is called pure node.

For example, the tree has only `Pass` is called pure node. No further splitting needed.



---



**- Gini Impurity**

It measures the how mixed the classes are.

For example, the tree with classes like `Pass` and `Fail`. If it has only `Pass`, then Gini = 0. Low gini.

If the data is low gini = Pure and High gini = Impure.



---



**- Entrophy**

Entrophy measures disorder or randomness.

Lower entropy means cleaner groups. Higher entropy means mixed groups.

We need to reduce entrophy, that means reducing the error.



---



**- Information Gain**

Information Gain means how much a question improves our knowledge. How much information we can get.

It is measured by:

Information Gain = Entropy (Before Split) − Entropy (After Split)

The information gain should be higher.



---



***VISUAL REPRESENTAION***

image.png



---



---



---
