# XGBOOST ALGORITHM

Problem Statement : We are going to classify breast cancer based on the given features using XGboost algorithm.

#### IMPORTING NECESSARY LIBRARIES

```python
# Data handling using numpy and pandas libraries
import numpy as np
import pandas as pd
# Importing train test split, XGBoost model and accuracy score
from sklearn.model_selection import train_test_split
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score
```

#### LOADING THE DATASET

```python
data = pd.read_csv("/content/drive/MyDrive/Boosting Algorithm/Breast Cancer Prediction.csv")    # Loading the dataset
data.head()                                                                                     # printing top 5 records
```

#### DATA PREPROCESSING

`Checking the null values present in the dataset.`

```python
data.isnull().sum()
```

As we already confirmed there are no null values present in the dataset.

```python
data.head()
```

`Checking the unique values in target column.`

```python
data['Class'].unique()
```

Only two classification(Class 2 and Class 4) available in target column.

`We need to convert class 2 and class 4 with 0 and 1 for easy understanding.`

```python
target_Class = []   # Empty list where storing the values in each iteration

for i in data['Class']:
  if i==2:
    print(0)
    target_Class.append(0)
  else:
    print(1)
    target_Class.append(1)
```

```python
target_Class  # Printing the new target class
```

```python
# Create a new column with the replaced target values.
data['Class_corrected'] = target_Class
```

```python
data.head()
```

#### SEGREGATING THE DATASET INTO INPUT AND OUTPUT

```python
x = data.iloc[:,1:-2].values
```

Segregate except target columns and conver to arrays.

```python
x
```

```python
y = data.iloc[:,-1].values
y
```

SPLITTING THE DATASET INTO TRAINING AND TESTING

```python
x_train, x_test, y_train,y_test = train_test_split(x,y, test_size= 0.2, random_state=42) # Splitting 80% training and 20% testing
```

```python
x_train.shape, x_test.shape, y_train.shape,y_test.shape
```

```python
x.shape, y.shape
```

LOADING THE XGBOOST MODEL

```python
model = XGBClassifier()
```

```python
model
```

Model imported and not trained.

TRAINING THE MODEL

```python
model.fit(x_train, y_train)
```

XGBoost Model is trained.

PREDICTING THE RESULT USING THE TRAINED MODEL

```python
model.predict(x_test)
```

```python
y_pred = model.predict(x_test)   # Model predicted value
y_pred
```

CALCULATING THE ACCURACY OF THE TRAINED MODEL

```python
accuracy_score(y_test, y_pred)    # Actual value Vs predicted value
```

The XGBoost model gives 94% accuracy which is pretty good for the given small dataset and XG boost performed well.
