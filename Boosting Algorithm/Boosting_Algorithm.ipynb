# BOOSTING ALGORITHM

What is boosting?

- Boosting is an ensemble learning technique that builds sequentially with each new learner improving the combined model by focusing on errors from the previous stages.

image.png

*Example*

For example, the teacher asks five students to solve difficult maths problems.

First student makes mistakes. The teacher correct that mistake, then the next learner tries solve better. The process continues, with each new student learning from what went wrong previously.

By the end, the group has built a much stronger solution. This is called boosting.

//////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////

**Difference between Bagging and Boosting?**

Bagging and Boosting are both ensemble learning techniques used to improve model performance by combining multiple models. The main difference is that:

- Bagging reduces variance by training models independently.

- Boosting reduces bias by training models sequentially, focusing on previous errors.

**How boosting works?**

*1. Build the first model*  - The first Decision Tree predicts the results of 10 students. It gets 7 correct and 3 wrong.


*2. Focus on the mistakes* - The next model pays more attention to the difficult students whom the first model classified incorrectly.

*3. Build the next model* - The second tree learns to handle some of those difficult cases. It may correct mistakes but could make other mistakes too.

*4. Continue improving* - More models are added sequentially. Each new model contributes to improving the combined prediction.

*5. Combine the models* - The final prediction combines the learners' contributions. The exact combination method depends on the boosting algorithm.

**Major Boosting Algorithms**

• AdaBoost (Adaptive Boosting): Focuses heavily on re-weighting misclassified data points and individual hypotheses after each iteration.

• Gradient Boosting: Trains each new model to fit the pseudo-residuals (errors) of the prior ensemble using gradient descent optimization.

• XGBoost (Extreme Gradient Boosting): An optimized, high-performance implementation of gradient boosting widely used in competitive data science.

• LightGBM : efficient gradient boosting, designed to scale well on many datasets.

• CatBoost : gradient boosting with particularly convenient handling of categorical features.

//////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////

Pros and Cons

• Pros:

	 - Effectively reduces bias and improves overall prediction accuracy.

	 - Handles complex non-linear relationships and diverse types of data.

• Cons:

	 - Computationally intensive because models must be trained sequentially rather than simultaneously.
   
	 - Sensitive to noise and outliers, which can be overemphasized during the re-weighting process.

**Important Boosting Terminologies**

*Weak Learner*

A weak learner is a model that performs a little better than random guessing on a particular learning task.

*Strong Learner*

A strong learner is a model that achieves better predictive performance by combining the contributions of multiple learners.

*Base Learner / Base Estimator*

A base learner is the individual model used as a building block in an ensemble.

*Sequential Learning*

In Boosting, models are built one after another, rather than all learning independently.

*Error / Misclassification*

An error occurs when the model's prediction differs from the actual answer.

*Sample Weight*

A sample weight indicates how much influence a training example has during learning.

*Residual*

A residual is the difference between the actual value and the predicted value.

Loss Function

A loss function measures how wrong a model's predictions are.

Examples:
- Regression: Mean Squared Error (MSE)
- Binary classification: Log Loss

*Learning Rate*

The learning rate controls how much each new learner contributes to the existing model.

*Number of Estimators*

The number of estimators is the number of boosting stages or learners used in the ensemble.
