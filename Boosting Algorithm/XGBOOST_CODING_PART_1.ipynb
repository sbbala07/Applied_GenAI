`Checking the unique values in target column.`

```python
data['Class'].unique()
```

Only two classification(Class 2 and Class 4) available in target column.

`We need to convert class 2 and class 4 with 0 and 1 for easy understanding.`

```python
target_Class = []   # Empty list where storing the values in each iteration

for i in data['Class']:
  if i==2:
    print(0)
    target_Class.append(0)
  else:
    print(1)
    target_Class.append(1)
```

```python
target_Class  # Printing the new target class
```

```python
# Create a new column with the replaced target values.
data['Class_corrected'] = target_Class
```
