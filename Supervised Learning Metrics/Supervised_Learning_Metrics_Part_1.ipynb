# SUPERVISED LEARNING METRICS

EVALUATION METRICS

- Evaluation metrics are measurements used to access how well our trained model makes predictions.

- It helps to understand our model makes good predictions or bad predictions.

`----------------------------------------------------------------------------------------------------------------`

`*Example*`

Imagine we are a teacher.

And we are conducting exam for 100 students. After correcting papers and announce:

"The class scored 90%".

Is it enough to understand how well every student performed?

No. We also want to know

- Howmany students passed?
- Howmany students failed?
- Howmany students who deserved to pass incorrectly marked as failed?
- Howmany students who failed incorrectly marked passed?

So, evaluation metrics is importance to understand the model predictions.


image.png

`----------------------------------------------------------------------------------------------------------------`

**Why do we need evaluation metrics?**

Imagine we are building a model to predict whether the customer will leave telecom company.

The model predicts

- 90% customers correctly
- 10% customers correctly.

We can say, our model has 90% accuracy.

But what if the model fails to identify most of the customers are actually going to leave?

The company may lose valuable customers.

That why we need evaluation metrics, one metric can't explain the every aspect of the model performance.


image.png

We use evalution metrics for two types of machine learning problems.

- Classification
- Regression

*Example*

`Classification`

Predicting a category or class

Examples:
- Email → Spam / Not Spam
- Customer → Will Leave / Will Stay
- Medical test → Positive / Negative
- Loan → Approved / Rejected


`Regression`

Predicting a numerical value

Examples:
- House price → ₹50 lakh
- Temperature → 32°C
- Sales → ₹5 lakh
- Gold price → ₹12,000 per gram

Let us understand the basic evaluation metrics used in classification and regression problems.

`----------------------------------------------------------------------------------------------------------------`

**ACCURACY SCORE**

- Accuracy score is the metrics used to find the relationship between the predicted value and the actual value in the classification problem.

Accuracy tells us how many predictions the model got correct out of all predictions.

Imagine we write an exam with 100 questions and answer 85 correctly.

Our accuracy is 85%.

Formula:

Accuracy= (TP+TN) / (TP+TN+FP+FN)


Imagine:

Accuracy= (35+50) / (100)

Accuracy=85%

Interpretation: The model correctly predicted the outcome for 85 out of 100.

We can use accuracy:
- When classes are reasonably balanced.
- When false positives and false negatives have similar importance.
- When you need a simple overall performance measure.

Limitation: Accuracy can be misleading when the dataset is imbalanced.

Example: Imagine 1,000 patients, of whom only 10 have a disease. A model that predicts everyone as healthy gets 99% accuracy, but fails to identify all 10 sick patients!

`----------------------------------------------------------------------------------------------------------------`

**R2 SCORE** - Coefficient of Determination

- R2 score is the metrics used to find the relationship between the predicted value and acutal value in the regression problem.


R² tells us:
How much of the variation in the actual target values is explained by the model?

Its formula is:

R² = 1-Sum of Squared Residuals / Total Sum of Squares



image.png

`Example`

Suppose our house-price model gives:

R² = 0.80

We can explain it as:
The model explains approximately 80% of the variation in house prices in the evaluated dataset.

The remaining 20% is not explained by the model.

- R² does not mean 80% of predictions are correct.

We can use R² when:

- Comparing regression models on the same test dataset.
- Understanding how much variation the model explains.
- Evaluating a baseline regression model.

Limitations:

- R² can be negative on test data.
- A high R² does not guarantee low practical prediction errors.
- It can increase when unnecessary features are added to an ordinary training regression model.
- It does not prove that the model has learned a causal relationship.

**Confusion Matrix**

A Confusion Matrix is a table that compares actual results with model predictions.

Let's assume the following results:
- 40 patients actually have the disease.
- 60 patients do not have the disease.
- The model correctly identifies 35 patients with the disease.
- It misses 5 patients with the disease.
- It incorrectly identifies 10 healthy patients as having the disease.
- It correctly identifies 50 healthy patients.

image.png

`True Positive (TP)`

   - The model predicts Positive, and the actual result is also Positive.

   - Example: A patient has a disease, and the model correctly identifies it.


`True Negative (TN)`

   - The model predicts Negative, and the actual result is also Negative.

   - Example: A patient is healthy, and the model correctly identifies that the patient does not have the disease.

`False Positive (FP)`

   - The model predicts Positive, but the actual result is Negative.

   - Example: The patient is healthy, but the model says the patient has a disease.

   - This is also called a Type I Error.

`False Negative (FN)`

   - The model predicts Negative, but the actual result is Positive.

   - Example: The patient has a disease, but the model says the patient is healthy.

   - This is also called a Type II Error.

`----------------------------------------------------------------------------------------------------------------`

**Precision**

Precision tells: Out of all the positive predictions made by the model, how many were actually positive?

Imagine the model identifies 45 patients as having a disease.

But only 35 really have the disease.

Precision measures how many of those 45 alerts were correct.

Formula:

Precision = TP / (TP+FP)


Using our example:

Precision = 35 / (35+10)

Precision = 77.78%


- Out of every 100 positive predictions, approximately 78 are actually positive.

When to use:

When false positives are expensive.

Examples:
- Spam detection: avoid marking genuine emails as spam.
- Fraud detection: avoid unnecessarily blocking genuine transactions.
- Medical diagnosis: avoid unnecessary alarming alerts and follow-up tests.

`----------------------------------------------------------------------------------------------------------------`

**Recall**

Recall tells : Out of all the actual positive cases, how many did the model successfully identify?

Imagine 40 patients actually have the disease.

The model identifies 35 correctly and misses 5.

Formula:

Recall = TP / (TP+FN)


Recall = 35 / (35+5)


Recall = 87.5%

- The model identifies 87.5% of all patients who actually have the disease.

When to use:

When missing a positive case is dangerous or expensive.

Examples:
- Cancer screening.
- Fraud detection.
- Identifying customers likely to leave.
- Detecting safety defects in manufacturing.

`----------------------------------------------------------------------------------------------------------------`

**F1-Score**

Sometimes we want both Precision and Recall to be good.

F1-Score combines these two metrics using their harmonic mean.

Formula:

F1 = 2 x (Precision x Recall) / (Precision+Recall)


example:


F1 = 2 x (0.7778 x 0.875) / (0.7778+0.875)


F1 = 82.35%


Interpretation: F1 provides a single measure that balances Precision and Recall.

When to use:
- Imbalanced classification datasets.
- Fraud detection.
- Spam detection.
- When both false positives and false negatives matter.

Limitation: F1 does not include True Negatives. It also does not tell the business cost of different mistakes.
