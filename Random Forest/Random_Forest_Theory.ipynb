## RANDOM FOREST

- Random Forest is a supervised machine learning algorithm that creates multiple decision trees and combines their results to make predictions.

- It is used for both classification and regression problems.

image.png

In simple words, instead of trusting single decision tree, Random forest creates multiple decision trees and take their combined results.

^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^

**Why do we use Random Forest Algorithm?**

Random forest is similar to decision tree algorithm, but it overcomes the limitations of Decision Tree.

The problem of decision tree is:

- It is overfitting
- The tree become too deep
- Memorize the training data instead of learning.

^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^

**How it will solve the above problems?**

If decision tree gives the predictions, but what is the prediction is a bad decision.

The Random forest creates multiple decision trees and getting all the predictions. Based on the majority voting/averaging, it gives the outcome. So the chance of giving bad decision in random tree is very less. Instead of relying on single tree, it gets result of multiple trees.

**Example of Random Forest**

For example, we want to select a student for cricket team.

Instead of asking one teached, we will ask 5 teacher to give the opinion.

Teacher 1 --> Yes

Teacher 2 --> No

Teacher 3 --> No

Teacher 4 --> Yes

Teacher 5 --> Yes


Here Yes - 3 and No - 2.

The majority is Yes. We will select the student as per the majority voting.

We use this method for classification problem.

For regression problem, use the same method and we will average the outcome and make the predicitions.

For example, if we want to predict house price.

Tree 1 → ₹50 lakh

Tree 2 → ₹55 lakh

Tree 3 → ₹52 lakh

Tree 4 → ₹48 lakh

Tree 5 → ₹50 lakh

Random Forest takes the average.

Prediction = 50 + 55 + 52 + 48 + 50 / 5
	​

Final prediction = ₹51 lakh

**Why is it Random Forest?**

- Random - It selects random data(Random sample of training data) and features randomly(Random subset of trees instead of all features) for each decision tree.

- Forest - A collection of trees.

**Bagging**

Bagging is otherwise called Bootstrap Aggregating.

Bootstrap - Random selection of data and features.

Aggregating - Combine the predictions.


Random forest uses bagging method to make predictions.

**Ensemble Learning**

Ensemble Learning means combining multiple models to make stronger model.

For example,

One student answer 70,

Other student answer 75,

Other student answer 80,

Instead of rely on one student, we will take combines opinion for reliable.

Random Forest is an example of ensemble learning.

**Advantages of Random Forest**

*- Reduces Overfitting*

*- It works for both classification and regression problems*

*- More accurate than decision tree*

*- Handles many features*

*- It can provide feature importance*

^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^

**Limitations of Random Forest**

*- Computationally more expensive*

*- It can still Overfit*

*- Larger model and can't visualize*
